```python
import pandas as pd

iraq_conflict = pd.read_csv("conflict_data_irq.csv")

iraq_conflict.head()
```

```python
iraq_conflict["total_deaths"] = (
    iraq_conflict["deaths_a"]
    + iraq_conflict["deaths_b"]
    + iraq_conflict["deaths_civilians"]
    + iraq_conflict["deaths_unknown"]
)

iraq_conflict[["deaths_a", "deaths_b", "deaths_civilians", "total_deaths"]].head()

iraq_conflict.to_csv("conflict_data_irq_totals.csv", index=False)

conflict_data = pd.read_csv("conflict_data_irq_totals.csv")

conflict_data.head(10)
```

```python
table1 = (
    iraq_conflict.groupby("year")["total_deaths"]
    .agg(
        average_deaths_per_event="mean",
        median_deaths_per_event="median",
        mode_deaths_per_event=lambda x: x.mode().iloc[0],
        total_deaths_year="sum",
    )
    .reset_index()
)
print(table1)
```

```python
import csv


def my_mean(values):
    total = 0
    for v in values:
        total += v
    return total / len(values)


deaths_by_year = {}

with open("conflict_data_irq_totals.csv", "r", newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        year = int(row["year"])
        total = int(row["total_deaths"])
        if year not in deaths_by_year:
            deaths_by_year[year] = []
        deaths_by_year[year].append(total)


def my_median(values):
    ordered = sorted(values)
    n = len(ordered)
    middle = n // 2
    if n % 2 == 1:
        return ordered[middle]
    return (ordered[middle - 1] + ordered[middle]) / 2


def my_mode(values):
    counts = {}
    for v in values:
        if v in counts:
            counts[v] += 1
        else:
            counts[v] = 1
    highest = max(counts.values())
    tied = [v for v, c in counts.items() if c == highest]
    return min(tied)


print(f"{'year':<6}{'mean':>10}{'median':>10}{'mode':>8}")
for year in sorted(deaths_by_year):
    d = deaths_by_year[year]
    print(f"{year:<6}{my_mean(d):>10.2f}{my_median(d):>10}{my_mode(d):>8}")
```
