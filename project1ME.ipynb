Project 1 : Middle East Analysis

```python
import pandas as pd

ME_conflict = pd.read_excel("Middle-East_aggregated_data_up_to_week_of-2026-10-31.xlsx")

ME_conflict.head()
```

```python
ME_conflict["year"] = pd.to_datetime(ME_conflict["WEEK"]).dt.year
```

```python
table1 = (
    ME_conflict.groupby(["year", "COUNTRY"])["FATALITIES"]
    .agg(
        average_deaths_per_event="mean",
        median_deaths_per_event="median",
        mode_deaths_per_event=lambda x: x.mode().iloc[0],
        total_deaths_year="sum",
    )
    .reset_index()
)
print(table1)
```
